## Introduction to linear regression
Linear regression is a statistical method used to model the relationship between a dependent variable and one or more independent variables by fitting a linear equation to the observed data. The primary purpose of linear regression is to predict the value of the dependent variable based on the independent variables or to understand the strength and direction of the relationship between them.


## The math

---
The basic mathematical formula behind simple linear regression, which involves one independent variable, is:
y = mx + b

In this equation:
- 'y' represents the dependent variable (the value you are trying to predict).
- 'x' represents the independent variable (the variable used for prediction).
- 'm' represents the slope of the line (the change in 'y' for a one-unit change in 'x').
- 'b' represents the y-intercept (the value of 'y' when 'x' is 0).

This is the formula for simple linear regression. For multiple independent variables, the concept is extended to multiple linear regression.



## Example data
Assume we have student Scores and the number of hours a student studies
|Id|Number of Hours|Marks Attained|
|--|---------------|--------------|
|S1|2|35|
|S2|3|45|
|S3|4|60|
|S4|6|75|
|S5|10|90|


Import the numpy library and create the small dataset as instructed.



```python
import numpy as np

X = np.array([2,3,4,6,10])
y = np.array([35,45,60,75,90])

print("Independent variable (X):", X)
print("Dependent variable (y):", y)
```

## Implement linear regression (manual)

### Subtask:
Show how to manually calculate the slope and intercept of the regression line using the example data.


**Reasoning**:
Calculate the mean of X and y, then calculate the sums required for the slope and intercept, and finally calculate and print the slope and intercept.



```python
mean_X = np.mean(X)
mean_y = np.mean(y)

deviation_product_sum = np.sum((X - mean_X) * (y - mean_y))
squared_deviation_sum_X = np.sum((X - mean_X)**2)

m = deviation_product_sum / squared_deviation_sum_X
b = mean_y - m * mean_X

print(f"Calculated slope (m): {m}")
print(f"Calculated intercept (b): {b}")
```

## Implement linear regression (library)

### Subtask:
Show how to use a library like scikit-learn to perform linear regression on the example data.


## Visualize the results


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(X, y, label='Data Points', color='blue')

predicted_y = m * X + b
plt.plot(X, predicted_y, label='Regression Line', color='red')

plt.xlabel("Independent Variable (X)")
plt.ylabel("Dependent Variable (y)")
plt.title("Linear Regression Example")
plt.legend()
plt.grid(True)
plt.show()
```

## Model Training

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X.reshape(-1, 1), y)

print(f"Scikit-learn slope (m): {model.coef_[0]}")
print(f"Scikit-learn intercept (b): {model.intercept_}")
```

## Interpret the results

### Subtask:
Interpret the meaning of the calculated slope and intercept in the context of the example data.


**Reasoning**:
Interpret the calculated slope and intercept and explain their meaning in the context of the linear regression formula.



```python
print(f"The calculated slope (m) is {m:.1f}. This means that for every one-unit increase in the independent variable (X), the dependent variable (y) is predicted to increase by {m:.1f} units.")
print(f"The calculated intercept (b) is {b:.1f}. This represents the predicted value of the dependent variable (y) when the independent variable (X) is 0.")
print(f"\nRelating this back to the linear regression formula (y = mx + b), for our example data, the fitted line can be represented as y = {m:.1f}x + {b:.1f}. This equation allows us to predict the value of 'y' for any given value of 'x' based on the relationship learned from the data.")
```
